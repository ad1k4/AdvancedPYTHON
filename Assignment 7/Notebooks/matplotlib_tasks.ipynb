```python
scores = [85, 78, 92, 88, 75]
names = ['Adilet', 'Arman', 'Sulu', 'Alisher', 'Aya']

plt.plot(names, scores)
plt.xlabel('Names')
plt.ylabel('Scores')
plt.title('Student Scores Line Plot')
plt.show()
```

```python
plt.bar(names, scores)
plt.xlabel('Names')
plt.ylabel('Scores')
plt.title('Student Scores Bar Chart')
plt.show()
```
